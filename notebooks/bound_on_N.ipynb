```python
epsilon = 0.5
C0 = 10-4*log(3)
C2 = 1/2+1/sqrt(2)
C3 = 1/2+1/sqrt(3)
bound_omega_sharper(N) = 1.3841*log(N)/log(log(N))
bound_omega(N) = N^epsilon
```

```python
g0(N) = -1/12 + 2^(-bound_omega(N))*(N/12 - sqrt(N)/(8*pi)*C2*C3*(2*log(N)+C0))
```

```python
B(N) = 2*bound_omega(N) + sqrt(N)/(2*pi)*(4*log(N)+C0)
```

```python
plot([g0,B], (N,6,400000))
```

```python
plot([bound_omega,bound_omega_sharper], (N,6,100))
```

```python
plot([g0term,g0term2,Bterm], (N,6,20000))
```
